# 02 - Inventory Analysis
Using the custom inventory optimization module to calculate EOQ, Reorder Points, and Safety Stock for specific products.

```python
import sys
import os
import pandas as pd

# Add parent directory to path so we can import our local modules
sys.path.append(os.path.abspath('..'))

from inventory.eoq import calculate_eoq
from inventory.reorder_point import calculate_reorder_point
from inventory.safety_stock import calculate_safety_stock
```

```python
# Assume we are analyzing 'WIDGET-01'
# Let's mock some annual metrics for this product
annual_demand = 12000  # units per year
ordering_cost = 75.0   # cost per PO
holding_cost = 1.50    # cost to hold 1 unit for a year

eoq_result = calculate_eoq(annual_demand, ordering_cost, holding_cost)
print(f"EOQ Result: {eoq_result}")

# Now calculate ROP
avg_daily_demand = annual_demand / 365
lead_time_days = 14
safety_stock = 100

rop_result = calculate_reorder_point(avg_daily_demand, lead_time_days, safety_stock)
print(f"ROP Result: {rop_result}")
```

```python
# Calculate dynamic safety stock based on a 95% service level
import numpy as np
demand_std_dev = np.std([10, 15, 12, 18, 9, 25, 11]) # Mock daily demand fluctuations

ss_result = calculate_safety_stock(avg_daily_demand, demand_std_dev, lead_time_days, service_level=0.95)
print(f"Safety Stock Result: {ss_result}")
```
