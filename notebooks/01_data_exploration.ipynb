# 01 - Data Exploration
This notebook is used to explore the raw supply chain data, understand data distributions, and identify missing values before running the ETL pipeline.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import os

sns.set_theme(style="whitegrid")
%matplotlib inline

RAW_DIR = '../data/raw'
```

```python
# Load raw data
sales_df = pd.read_csv(os.path.join(RAW_DIR, 'sales.csv'))
inventory_df = pd.read_csv(os.path.join(RAW_DIR, 'inventory.csv'))
products_df = pd.read_csv(os.path.join(RAW_DIR, 'products.csv'))

print(f'Sales records: {len(sales_df)}')
sales_df.head()
```

```python
# Check for missing values
print("Missing Values in Sales Data:")
print(sales_df.isnull().sum())

print("\nMissing Values in Inventory Data:")
print(inventory_df.isnull().sum())
```

```python
# Visualize sales quantity distribution
plt.figure(figsize=(10, 5))
sns.histplot(sales_df['quantity'], bins=30, kde=True)
plt.title('Distribution of Sales Quantities')
plt.xlabel('Quantity Sold')
plt.ylabel('Frequency')
plt.show()
```
