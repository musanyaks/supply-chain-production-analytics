# 03 - Demand Forecasting
Using Scikit-Learn to engineer time-series features, train a demand forecasting model, and visualize future predictions.

```python
import sys
import os
import pandas as pd
import matplotlib.pyplot as plt

sys.path.append(os.path.abspath('..'))
from forecasting.feature_engineering import create_time_features
from forecasting.train_model import train_demand_model
from forecasting.demand_forecasting import forecast_future_demand

# Load sales data
sales_df = pd.read_csv('../data/raw/sales.csv')
sales_df['sale_date'] = pd.to_datetime(sales_df['sale_date'])
sales_df.head()
```

```python
# Feature Engineering
# Let's pick a specific product to forecast
product_id = sales_df['product_id'].iloc[0]
prod_sales = sales_df[sales_df['product_id'] == product_id].copy()

featured_df = create_time_features(prod_sales, date_col='sale_date')
featured_df.head()
```

```python
# Train the model
features = ['dayofweek', 'quarter', 'month', 'year', 'dayofyear', 'lag_7', 'rolling_mean_7']
model = train_demand_model(featured_df, features, target='quantity')
print("Model trained successfully!")
```

```python
# Forecast future demand
future_forecast = forecast_future_demand(sales_df, product_id, days_ahead=30)

# Plot results
plt.figure(figsize=(12, 6))
plt.plot(prod_sales['sale_date'], prod_sales['quantity'], label='Historical Sales')
plt.plot(future_forecast['sale_date'], future_forecast['predicted_demand'], label='Forecasted Demand', color='red')
plt.title(f'Demand Forecast for Product {product_id}')
plt.xlabel('Date')
plt.ylabel('Quantity')
plt.legend()
plt.show()
```
