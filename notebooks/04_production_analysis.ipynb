# 04 - Production Analysis
Analyzing production efficiency (OEE) and identifying the primary causes of machine downtime using Pareto analysis.

```python
import sys
import os
import pandas as pd
import matplotlib.pyplot as plt

sys.path.append(os.path.abspath('..'))
from production.production_efficiency import calculate_oee
from production.downtime_analysis import analyze_downtime
```

```python
# Calculate OEE for a specific machine or line
# Availability = 88%, Performance = 95%, Quality = 99%
oee_result = calculate_oee(availability=0.88, performance=0.95, quality=0.99)
print(f"OEE Score: {oee_result['oee_percentage']}%")
print(f"Availability Loss: {oee_result['availability_loss']}%")
```

```python
# Load mock downtime data
downtime_data = {
    'reason_code': ['Machine Jam', 'Material Shortage', 'Operator Break', 'Sensor Failure', 'Machine Jam', 'Power Outage'],
    'downtime_hours': [12.5, 4.0, 8.0, 2.5, 15.0, 1.0]
}
downtime_df = pd.DataFrame(downtime_data)

# Run Pareto analysis
pareto_summary = analyze_downtime(downtime_df)
pareto_summary
```

```python
# Plot Pareto Chart
fig, ax1 = plt.subplots(figsize=(10, 5))

color = 'tab:blue'
ax1.bar(pareto_summary['reason_code'], pareto_summary['downtime_hours'], color=color)
ax1.set_xlabel('Downtime Reason')
ax1.set_ylabel('Total Downtime Hours', color=color)
ax1.tick_params(axis='y', labelcolor=color)
plt.xticks(rotation=45)

ax2 = ax1.twinx()
color = 'tab:red'
ax2.plot(pareto_summary['reason_code'], pareto_summary['cumulative_percentage'], color=color, marker='D', ms=5)
ax2.set_ylabel('Cumulative Percentage', color=color)
ax2.tick_params(axis='y', labelcolor=color)
ax2.axhline(80, color='gray', linestyle='--')

plt.title('Pareto Chart of Downtime Reasons')
fig.tight_layout()
plt.show()
```
